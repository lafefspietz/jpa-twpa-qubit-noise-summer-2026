```python
import pyvisa
import json
```

```python
rm = pyvisa.ResourceManager()
instruments = rm.list_resources()
instruments
```

```python
yoko = None
for instrument in instruments:
    if 'YOKOGAWA' in instrument:
        # Initialize instrument
        yoko = rm.open_resource(instrument) 
        break
print(yoko.query('*IDN?').strip())
```

```python
yoko.write(":SOUR:LEV 0.69")
```

```python
print(yoko.query(":SOUR:RANG?"))
```

```python
yoko_state = {}
yoko.write(":SOUR:FUNC VOLT")
yoko_state['units'] = "volts"
yoko_state['level'] = float(yoko.query(":SOUR:LEV?"))
yoko_state['on'] = True

with open("yoko_state.json", "w", encoding="utf-8") as f:
    json.dump(yoko_state, f, indent=4)
    
#voltage_level = yoko.query(":SOUR:LEV?")
#yoko.write(":SOUR:LEV 0.69")
#voltage_level = yoko.query(":SOUR:LEV?")
#yoko.write(":SOUR:RANG 1")
#voltage_range = yoko.query(":SOUR:RANG?")
#yoko.write(":OUTP ON")
#yoko.write(":OUTP OFF")
#output_state = yoko.query(":OUTP?")
#yoko.write(":SOUR:FUNC VOLT")
#yoko.write(":SOUR:FUNC CURR")
#source_mode = yoko.query(":SOUR:FUNC?")
```
