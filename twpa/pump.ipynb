```python
# pip install windfreak
import json
import serial.tools.list_ports
from windfreak import SynthHD
import time
ports = serial.tools.list_ports.comports()
for port in ports:
    print(f"Port Name:   {port.device}")
    print(f"Description: {port.description}")
    print(f"Hardware ID: {port.hwid}")  # Shows USB Vendor ID, Product ID, and Location string
    print("-" * 30)
    if 'A3E5' in port.hwid:
        print(f"Windfreak is on COM port {port.device}")

synth = SynthHD("COM4")
synth.init()
channel_a = synth[0]
channel_a.enable = False
channel_b = synth[1]
channel_b.enable = False
```

```python
channel_b.enable = True # turn channel B on
channel_b.frequency = 8.75e9   # frequency in Hz
channel_b.power = -25        #  power in dBm
```

```python
channel_b.enable = False # turn channel B off
```

```python
synth.close() # close here or it will hold the COM port and not let other programs use it
```
